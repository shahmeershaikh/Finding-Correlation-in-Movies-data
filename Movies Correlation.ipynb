```python
# Import Libraries

import pandas as pd
import seaborn as sns
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
#Adjust The Configuration Of The Plot
matplotlib.rcParams['figure.figsize'] = (12,8)

# Read In The Data
df = pd.read_csv (r'C:\Users\shahm\OneDrive\Desktop\Data Analysis Project\movies.csv')
```

```python
#Looking At The Data
df.head()
```

```python
# See If There Is Any Missing Data

for col in df.columns:
        percentage_missing = np.mean(df[col].isnull())
        print('{} - {}%' .format(col,percentage_missing))
```

```python
# Drop all rows with nulls
df = df.dropna()
```

```python
# Data Types For Columns
df.dtypes
```

```python
# Change Data Types For Columns

df['votes'] = df['votes'].astype('int64')
df['budget'] = df['budget'].astype('int64')
df['gross'] = df['gross'].astype('int64')
df['runtime'] = df['runtime'].astype('int64')
```

```python
df.head()
```

```python
df = df.sort_values(by=['gross'],inplace=False,ascending=False)
```

```python
pd.set_option('display.max_rows', None)
```

```python
df.sort_values(by=['gross'],inplace=False,ascending=False).head()
```

```python
# Drop Duplicates
df['company'].drop_duplicates().sort_values(ascending=False)
```

```python
df.drop_duplicates()
```

```python
df.head()
```

```python
# Compare Budget Vs Gross

plt.scatter(x=df['budget'],y=df['gross'])
plt.title('Budget Vs Gross Earning')
plt.xlabel('Gross Earning')
plt.ylabel('Budget For Film')
plt.show()
```

```python
df.head()
```

```python
# Plot Budget Vs Gross Using Seaborn

sns.regplot (x='budget', y='gross',data=df, scatter_kws={'color':'green'}, line_kws={'color':'red'})
```

```python
# Looking At Correlation

df.corr(numeric_only=True, method='pearson')
```

```python
# High Correlation Between Budget And Gross

correlation_matrix = df.corr(numeric_only=True, method='pearson')

sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Mattix Numeric')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')
plt.show()
```

```python
df_numerized = df

for col_name in df_numerized.columns:
    if(df_numerized[col_name].dtype == 'object'):
        df_numerized[col_name] = df_numerized[col_name].astype('category')
        df_numerized[col_name] = df_numerized[col_name].cat.codes
df_numerized.head()
```

```python
correlation_mat = df_numerized.corr()
corr_pairs = correlation_mat.unstack()
corr_pairs
```

```python
sorted_pairs = corr_pairs.sort_values()
sorted_pairs
```

```python
high_correlation = sorted_pairs[(sorted_pairs) > 0.5]
high_correlation
```
